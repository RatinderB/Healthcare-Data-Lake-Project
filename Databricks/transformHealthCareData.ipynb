```python
# Loop through each file in 'file_name' and load it as a DataFrame
for file in file_name:
    # Use 'exec' to create a DataFrame for each file based on its name
    exec(f"{file.split('.')[0]}_df = spark.read.parquet('/mnt/healthParquet/{file}')")
```

```python
# Function to apply schema conversion on the dataframe
def apply_schema(df, target_schema):
    # Loop through each field in the target schema
    for field in target_schema.fields:
        column_name = field.name  # Column name from the schema
        target_type = field.dataType  # Desired data type for the column

        # Check if the column exists in the dataframe and whether it needs casting
        if column_name in df.columns:
            current_type = dict(df.dtypes).get(column_name)
            if current_type != str(target_type):
                # If current type differs from the target, cast the column to the target type
                df = df.withColumn(column_name, col(column_name).cast(target_type))
    
    return df

# Iterate over each file name in the file_name list
for file in file_name:
    # Generate the schema variable name dynamically based on the file name (without the extension)
    schema_name = file.split(".")[0] + "_schema"
    
    # Access the actual schema object from the global namespace using globals()
    target_schema = globals().get(schema_name)

    # Generate the dataframe name dynamically based on the file name (without the extension)
    df_name = file.split(".")[0] + "_df"
    
    # Retrieve the dataframe object from the global namespace
    df = globals().get(df_name)
    
    # Apply the schema transformation only if both the schema and dataframe exist
    if target_schema and df:  
        globals()[df_name] = apply_schema(df, target_schema)
```

```python
def fill_missing_strings(df):
    # Identify string columns dynamically
    string_cols = [col_name for col_name, dtype in df.dtypes if "string" in dtype]
    
    # Fill NULLs in string columns with "Unknown" in one operation
    return df.fillna("Unknown", subset=string_cols)


for file in file_name:
    # Generate the dataframe name dynamically based on the file name (without the extension)
    df_name = file.split(".")[0] + "_df"
    
    # Retrieve the dataframe object from the global namespace
    df = globals().get(df_name)

    globals()[df_name] = fill_missing_strings(df)
    
```

```python
primary_keys = {
    "allergies_df": ["START", "STOP", "PATIENT", "ENCOUNTER", "CODE"],
    "careplans_df": ["Id"],
    "claims_df": ["Id"],
    "claims_transactions_df": ["ID"],
    "conditions_df": ["START", "STOP", "PATIENT", "ENCOUNTER", "CODE"],
    "devices_df": ["START", "STOP", "PATIENT", "ENCOUNTER", "CODE"],
    "encounters_df": ["Id"],
    "imaging_studies_df": ["Id"],
    "immunizations_df": ["DATE", "PATIENT", "ENCOUNTER", "CODE"],
    "medications_df": ["START", "STOP", "PATIENT", "ENCOUNTER", "CODE"],
    "observations_df": ["DATE", "PATIENT", "ENCOUNTER", "CODE"],
    "organizations_df": ["Id"],
    "patients_df": ["Id"],
    "payer_transitions_df": ["PATIENT", "MEMBERID", "START_DATE", "END_DATE"],
    "payers_df": ["Id"],
    "procedures_df": ["START", "STOP", "PATIENT", "ENCOUNTER", "CODE"],
    "providers_df": ["Id"],
    "supplies_df": ["DATE", "PATIENT", "ENCOUNTER", "CODE"],
}

# Iterate over all DataFrames and drop duplicates based on primary keys
for df_name, key_columns in primary_keys.items():
    df = globals().get(df_name)
    if df and key_columns:  # Ensure DataFrame exists and has a valid primary key
        globals()[df_name] = df.dropDuplicates(key_columns)

        # Cache the DataFrame for faster future processing
        globals()[df_name] = df.cache()
```

```python
patients_df = patients_df.withColumn("AGE", (datediff(current_date(), col("BIRTHDATE")) / 365).cast("int"))
```

```python
encounters_df = encounters_df.withColumn("OUT_OF_POCKET", expr("TOTAL_CLAIM_COST - PAYER_COVERAGE"))
```

```python
for file in file_name:
    # Generate the dataframe name dynamically based on the file name (without the extension)
    df_name = file.split(".")[0] + "_df"
    
    # Retrieve the dataframe object from the global namespace
    df = globals().get(df_name)

    # Check if the dataframe exists in globals() to avoid errors
    if df:
        # Define the Delta table path in the new healthdelta container
        delta_table_path = f"/mnt/healthDelta/{file.split('.')[0]}"

        # Write the DataFrame into the Delta table in the new healthDelta container
        df.write.format("delta").mode("overwrite").save(delta_table_path)
        print(f"DataFrame '{df_name}' written to Delta table at {delta_table_path}")
```

```python
# Dictionary mapping file names to their respective partition columns
partition_columns = {
    "claims": ["PATIENTID", "SERVICEDATE"],
    "claims_transactions": ["PATIENTID", "FROMDATE"],
    "conditions": "PATIENT",
    "encounters": "PATIENT",
    "imaging_studies": "PATIENT",
    "immunizations": "PATIENT",
    "medications": "PATIENT",
    "observations": "PATIENT",
    "payer_transitions": "PATIENT",
    "procedures": "PATIENT",
    "supplies": "PATIENT"
}

for file in file_name:
    df_name = file.split(".")[0] + "_df"
    df = globals().get(df_name)

    if df:
        delta_table_path = f"/mnt/healthDelta/{file.split('.')[0]}"
        
        # Determine partition columns
        partition_col = partition_columns.get(file.split('.')[0], None)
        
        if partition_col:
            if isinstance(partition_col, list):
                df.write.format("delta").mode("overwrite").partitionBy(*partition_col).save(delta_table_path)
            else:
                df.write.format("delta").mode("overwrite").partitionBy(partition_col).save(delta_table_path)
            print(f"DataFrame '{df_name}' written to Delta table at {delta_table_path} with partitioning on {partition_col}")
        else:
            df.write.format("delta").mode("overwrite").save(delta_table_path)
            print(f"DataFrame '{df_name}' written to Delta table at {delta_table_path} without partitioning")
```
